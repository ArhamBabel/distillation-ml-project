```python
import pandas as pd, numpy as np

new = pd.read_csv("../data/distillation_purity_dataset.csv")
old = pd.read_csv("../data/distillation_purity_dataset_5000_archive_old_energy.csv")

cols = ["z_F", "x_D", "x_B", "alpha", "R", "N", "system_id", "F", "lam"]
print("Inputs identical:", np.allclose(new[cols], old[cols]))

D_over_F = (new["z_F"] - new["x_B"]) / (new["x_D"] - new["x_B"])
print("Q ratio equals D/F:", np.allclose(new["Q_reb"] / old["Q_reb"], D_over_F))

print("D/F strictly between 0 and 1:", ((D_over_F > 0) & (D_over_F < 1)).all())
print("Q_reb all positive:", (new["Q_reb"] > 0).all())
```
